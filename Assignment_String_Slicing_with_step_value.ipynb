## To slice two sub strings ie. username and domain name from a single string with the help of count

```python
file = "myfile.pdf"
count = file.index('.')
print(count)
username = file[ : count: +1]
print("username :", username)
domain = file[count+1 ::+1]
print("domain :", domain)
```

```python
email = "pohareashini100@gmail.com"

count= email.index("@")


print(count)

username=email[ :count : +1]
print(username)
domain= email[count+1::+1]
print("domain", domain)
```

## accessing sub string w.t.h of step value
