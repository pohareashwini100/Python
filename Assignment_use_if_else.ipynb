
#### 1. Find number is Positive, Negative or Zero



```python
a = 10

if a > 0:
    print("Positive")
elif a < 0:
    print("Negative")
else:
    print("Zero")

```

#### 2. Find Largest of Two Numbers



```python
a = 20
b = 15

if a > b:
    print("a is greater")
elif b > a:
    print("b is greater")
else:
    print("Both are equal")

```
