### String Slicing

```python
name = "Ashwini"
print(name[0:5])
print(name[-1:-5])   # doesn't print anything only left blank line 
print(name[-3:-1])
print(name[0:9])       # print all string
print(name[0:100])       # print all string
print(name[-1:4])   # doesn't print anything only left blank line 
print(name[6:4])     # doesn't print anything only left blank line 
print(name[-5:-1])
print(name[6:-4])   # doesn't print anything only left blank line 
```
