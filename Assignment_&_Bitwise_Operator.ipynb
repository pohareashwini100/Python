# Bitwise Operator
## &  and , |  or , ^  xor , !  xnor, >> increment , << decrement
### We can represent integers in 4 ways
#### 1] Decimal 2] Binary 3] Octal 4] Hexadecimal

```python
x = 10

print(bin(x))  # decimal to binary

print(oct(x))  # decimal to octal

print(hex(x))  # decimal to hexa

print(type(x))   # type of x



x = "1010"
print(int(x, 2))  # binary to decimal

x = "1010"
print(int(x, 8))   # octal to decimal

x = "1010"
print(int(x, 16))   # hexadecimal to decimal


```

### to convert binary to octal
#### convert  binary → decimal, then decimal → octal.

### to convert binary to hexadecimal
#### convert  binary → decimal, then decimal → hexadecimal.

```python
x = "1010"

decimal = int(x, 2)
octal = oct(decimal)

print(octal)     # binary to octal


x = "1010"

decimal = int(x, 2)
hexa = hex(decimal)

print(hexa)      # binary to hexadecimal


x = "12"

decimal = int(x, 8)
binary = bin(decimal)

print(binary)   # octal to binary


x = "A"

decimal = int(x, 16)
binary = bin(decimal)

print(binary)    # hexa to binary
```
